# **randint() ga oid masalalar (1–5)**

1 dan 10 gacha bo‘lgan tasodifiy bitta sonni chiqaruvchi dastur yozing.

```python
from numpy import random
import numpy as np
```

5 ta elementdan iborat, 0 dan 50 gacha tasodifiy sonlardan iborat massiv yarating.

```python
import random
random_array = [random.randint(0, 50) for _ in range(5)]
print(random_array)
```

10 ta tasodifiy son (1–100 oralig‘ida) hosil qiling va ularning yig‘indisini toping.

```python
import random
tasodifiy_sonlar = [random.randint(1, 100) for _ in range(10)]
yigindi = sum(tasodifiy_sonlar)

print(f"Tasodifiy sonlar: {tasodifiy_sonlar}")
print(f"Ularning yig'indisi: {yigindi}")
```

3x3 o‘lchamli matritsa yarating (elementlari 0–9 oralig‘ida bo‘lsin).

```python
arr = random.randint(0,9, size = (3,3))
arr
```

20 ta tasodifiy son ichidan faqat juft sonlarni ajratib chiqaring.

```python
arr = random.randint(0,100, size = (20))
arr
```
